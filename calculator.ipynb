# Basic Calculator with Execution Time

Addition, subtraction, multiplication and division, with the execution time of each calculation.

Run the cells from top to bottom.

## 1. Define the operations

```python
import time


def add(a, b):
    return a + b


def subtract(a, b):
    return a - b


def multiply(a, b):
    return a * b


def divide(a, b):
    if b == 0:
        raise ZeroDivisionError("Cannot divide by zero.")
    return a / b
```

## 2. Run an operation and measure its execution time

```python
def calculate(func, a, b):
    start = time.perf_counter()
    result = func(a, b)
    elapsed_ms = (time.perf_counter() - start) * 1000
    return result, elapsed_ms


for name, symbol, func in [
    ("Addition", "+", add),
    ("Subtraction", "-", subtract),
    ("Multiplication", "*", multiply),
    ("Division", "/", divide),
]:
    result, ms = calculate(func, 20, 4)
    print(f"{name}: 20 {symbol} 4 = {result}  |  Execution time: {ms:.6f} ms")
```

## 3. Interactive calculator

Run this cell and type your inputs in the box that appears. Choose `5` to exit.

```python
OPERATIONS = {
    "1": ("Addition", "+", add),
    "2": ("Subtraction", "-", subtract),
    "3": ("Multiplication", "*", multiply),
    "4": ("Division", "/", divide),
}


def get_number(prompt):
    while True:
        try:
            return float(input(prompt))
        except ValueError:
            print("Invalid input. Please enter a number.")


while True:
    print("\n===== Basic Calculator =====")
    for key, (name, symbol, _) in OPERATIONS.items():
        print(f"{key}. {name} ({symbol})")
    print("5. Exit")

    choice = input("Choose an operation (1-5): ").strip()

    if choice == "5":
        print("Goodbye!")
        break
    if choice not in OPERATIONS:
        print("Invalid choice. Please select 1-5.")
        continue

    name, symbol, func = OPERATIONS[choice]
    a = get_number("Enter first number: ")
    b = get_number("Enter second number: ")

    try:
        result, ms = calculate(func, a, b)
    except ZeroDivisionError as e:
        print(f"Error: {e}")
        continue

    print(f"\n{name}: {a} {symbol} {b} = {result}")
    print(f"Execution time: {ms:.6f} ms")
```
